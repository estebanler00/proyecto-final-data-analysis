```python
#La cantidad de nulos restantes en el DB representa un 0,01% del total de registros(1300 nulos / +130000 registros totales), por lo que se decide dropearlos sin representar una perdida significativa de información.

df1_clean.dropna(inplace=True)

df1_clean.isnull().sum()
```

```python
#Verificado que los vlaores de los registros de la DB son válidos, se procede a hacer el recorte de los registros para tener un volumen de información menos pesado y más manipulable para realizar el análisis.
#Se toman como criterios de recorte los factores determinantes para el éxito de un videojuego. EN este sentido, recortaremos los registros con menor cantidad de ventas (expresadaas en la columna estimated_owners).

df1_clean.estimated_owners.value_counts()
```

```python
#Al verificar los valores por cada rango de ventas, vemos que los valores tienen diferencia de sintaxis, por lo que se debe reemplazar las "," y los ".." que hacen que los registros con la misma cantidad de ventas se vean representados en distintos grupos.

df1_clean["estimated_owners"] = df1_clean.estimated_owners.str.replace(",", "").str.replace("..", "-")
```

```python
df1_clean.estimated_owners.value_counts()
```

```python
#Normalizados los valores de estimated_owners, ahora si procedemos a recortar los registros que estén por debajo de los 20000 ventas.

df1_clean = df1_clean.drop(df1_clean[df1_clean.estimated_owners == "0 - 20000"].index)
df1_clean = df1_clean.drop(df1_clean[df1_clean.estimated_owners == "0 - 0"].index)
```

```python
#Conexión a la nueva DB limpia para continuar con el tratamiento de duplicados y normalización de los datos.

df_clean = pd.read_csv("C:/Users/esteb/Documents/Esteban/CURSOS/Data-analisis-ComIT_ok/proyecto-final-data-analysis/data/steam_games_clean.csv")
```

```python
# Normalización de formato de fechas a partir de la columna release_date

patron_fecha = r"^\d{4}-\d{2}-\d{2}$"

formato_correcto = df_clean["release_date"].astype(str).str.match(patron_fecha)

fechas_convertidas = pd.to_datetime(
    df_clean["release_date"],
    format="%Y-%m-%d",
    errors="coerce"
)

fecha_valida = fechas_convertidas.notna()

fechas_invalidas = df_clean[~(formato_correcto & fecha_valida)]

fechas_invalidas["release_date"].count()

#No se encuentran errores en el formato de las fechas.
```

```python
#Con los valores de las fechas validados, verificamos que las fechas de lanzamiento sean congruentes con el lanzamiento de Steam. Evitando que haya algún juego con fecha de lanzamiento muy antigua.

df_clean.release_date[df_clean.release_date < "1995-01-01"]

#No se encuentran fechas fuera de lugar.
```

```python
#Control de valores negativos en la columna de precios. No hay valores negativos.

df_clean.price[df_clean.price < 0].count()
```

```python
#Verificación de duplicados en la DB. Normaización de nombres y detección de duplicados a partir de fecha y nombre. 28 duplicados detectados.

df_clean["name"] = df_clean.name.str.lower()

duplicados = df_clean[df_clean.duplicated(subset=["name", "release_date"], keep=False)]

duplicados.count()


```

```python
#Eliminación de duplicados.

df_clean.drop_duplicates(subset=["name", "release_date"], keep="first", inplace=True)

#Verificación de duplicados tras la eliminación de los mismos.

duplicados.count()
```
